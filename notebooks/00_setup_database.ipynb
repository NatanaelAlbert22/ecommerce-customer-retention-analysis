```python
from pathlib import Path
import duckdb

# Root proyek = satu folder di atas folder notebooks/
PROJECT_ROOT = Path.cwd().parent
DATA_DIR = PROJECT_ROOT / "data"

con = duckdb.connect(str(DATA_DIR / "olist.duckdb"))
```

```python
tables = [
    "olist_customers_dataset",
    "olist_orders_dataset",
    "olist_order_items_dataset",
    "olist_products_dataset",
    "olist_sellers_dataset",
    "olist_order_payments_dataset",
    "olist_order_reviews_dataset",
    "olist_geolocation_dataset",
    "product_category_name_translation",
]

for t in tables:
    csv_path = DATA_DIR / "raw" / f"{t}.csv"
    con.execute(f"""
        CREATE OR REPLACE TABLE {t} AS
        SELECT * FROM read_csv_auto('{csv_path.as_posix()}')
    """)
    print(t, "loaded:", con.execute(f"SELECT COUNT(*) FROM {t}").fetchone())
```
